Understand what you want to accomplish. This is the first step of the CRISP-DM framework for data science projects. Business understanding is about looking at the data in context. Skipping this step could lead to projects that are intensive work but lack any real meaning or impact in the world the data came from or in the organization looking for insight from it.

**a. Read the dataset description. Who produced tha data? When, how, and for what purpose?**

**Answer:** Both datasets are published by the Tartu City Government on the Tartu open data portal (ArcGIS Hub). Dataset 1 (about 14 GB, roughly 70 million rows) is collected automatically by the radar speed signs (“Sinu kiirus”) on Tartu streets. Each row appears to be one passing vehicle, with the date and time, the sign ID and its measured speed. The data covers about 2020 to July 2026. Dataset 2 (4 KB) describes the signs themselves: their location, the speed limit at that spot and their status. The signs are meant to calm traffic by showing drivers their speed, and the city uses the recorded data to monitor traffic volume and speeds at these locations.

**b. Name a plausible stakeholder, their goals, and the concrete decision the analysis would support. Who benefits, who could be harmed?**

**Answer:** The main stakeholder is the Tartu City Government’s traffic management department. Its goal is to reduce speeding and improve road safety with a limited budget. The analysis supports decisions on where and when to add traffic-calming measures (speed bumps, raised crossings, speed cameras, lower limits) and where to move the signs next. The municipal police could use the results to schedule speed checks at high-risk times. Residents, pedestrians, cyclists and children near schools benefit. Drivers on streets flagged as high-risk may face more checks and fines. Areas without signs may get less attention, because the data only covers places where signs already stand.

**c. State the objective in business terms (decide strategy, predict demand or delay, recognize priorities, connect factors to outcomes, detect risks, etc), and how success would be judged (cost saved, error tolerated, etc).**

For example:
* _"Predict delay for bus lines so that passengers can choose an earlier connection."_
* _"Recognize priorities among customer complaints so that support staff answer urgent ones first."_

NB! For this step, remember to name the objective without naming the method. For example, don't state “predict X with a random forest” as the objective.

**Answer:** Find out on which days, at which times, in which months and in which areas speeding is most common and most severe, understand which conditions are linked to speeding (speed limit, type of road, weather, day of the week), and identify high-risk hours in advance, so the city can target traffic-calming measures and police checks where they reduce speeding the most. Success means:

* the high-risk locations and times we find are consistent across different months and agree with the city’s own knowledge of problem streets;
* the high-risk hours are identified reliably enough to plan checks, for example most actual high-speeding hours are caught without too many false alarms;
* after measures are introduced at the chosen locations, the share of speeding vehicles there decreases.


**d. Now translate the objective to a data science problem.**

Is it classification, regression, clustering, description? What would be predicted or explained? At what point in time would the prediction or other outcome be used?

**Answer:** The project combines three types of problems:

1) **Descriptive analysis (Goal 1)**: aggregate the data by sign, hour, weekday and month, and compare the share of speeding vehicles and how far above the limit they go.
2) **Clustering (Goal 2)**: group signs or sign-hours by their speeding pattern and conditions (speed limit, location, road type, weather), then check whether the clusters match our hypotheses.
3) **Binary classification (Goal 3)**: for a given sign and hour, predict whether the average speed will exceed the speed limit. Inputs are the sign and its speed limit, hour, weekday, month, traffic volume and possibly weather.

The prediction would be used ahead of time, when the city or police plans next week’s checks or the next period’s traffic-calming measures.

**e. Identify potential constraints.**

These could be time limitations, data privacy concerns, regulatory requirements, technical limitations, and ethical considerations.

**Answer:** 
* **Data size**: about 14 GB, and the API returns at most 1000 rows per query. We have to aggregate on the server or in chunks (for example to hourly level) instead of working with all raw rows at once.
* **Few signs**: Tartu has only a small number of signs, so the data is repetitive. Models may learn the specific signs rather than general conditions, which limits how well results apply to other streets or cities.
* **Data quality**: some timestamps are clearly wrong (dates in 2000 from reset device clocks). The meaning of the speed fields is not fully documented, and signs that moved or changed status have to be matched correctly with dataset 2.
* **Measurement bias**: the signs are not calibrated enforcement radars, so the data cannot be used as legal evidence. The signs themselves change driver behaviour, and they are placed on selected streets, so they do not represent the whole city.
* **Extra data**: weather or data from other cities would have to be found and joined by time and location, which may not be available at the same level of detail.
* **Privacy and ethics**: the data has no personal information such as licence plates, so privacy risk is low. The results could still lead to stricter enforcement in some neighbourhoods, which should be justified by safety and not by fine revenue.
* **Time**: the project has to be finished within the semester alongside other courses.

**c. First shots at visualizing the data**

Take a look at the data as plots to get more insight to the patterns, trends, and relationships. This could be histograms, scatter plots, box plots, ect. What did you gather? What stands out?

```python
import pandas as pd
import requests

url = "https://gis.tartulv.ee/arcgis/rest/services/Hosted/UUS_kiirustablood_Kuvamine/FeatureServer/0/query"

data = requests.get(url, params={
    "where": "1=1",
    "resultRecordCount": 100000,
    "outFields": "*",
    "f": "json"
}).json()

rows = []

for x in data["features"]:
    rows.append(x["attributes"])

df = pd.DataFrame(rows)

df.head()
```

```python
import matplotlib.pyplot as plt

plt.hist(df["V1"].dropna(), bins=20)
plt.xlabel("Speed (km/h)")
plt.ylabel("Number of measurements")
plt.title("Speed distribution")
plt.show()
```

```python
plt.scatter(df["V1"], df["V2"])

plt.xlabel("V1 (km/h)")
plt.ylabel("V2 (km/h)")
plt.title("V1 vs V2")

plt.show()
```

### Observations

The histogram shows that most V1 speed measurements are between 40 and 65 km/h, with the highest frequency around 50–55 km/h. There are also a few unusually high and low values.

The scatter plot shows a positive relationship between V1 and V2. As V1 increases, V2 generally increases as well.

We only analyzed a sample from one radar, these observations cannot be generalized to all radar locations in Tartu.
