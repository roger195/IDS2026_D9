# Kiirustabloode DB

```python
import pandas as pd

df = pd.read_csv("Data/TV_kiirustabloo.csv")
df.head()
```

## Loome database kiirustabloodele

```python
import duckdb

con = duckdb.connect("md:")

con.execute("CREATE DATABASE IF NOT EXISTS radarid_db")

con.execute("""
    CREATE OR REPLACE TABLE radarid_db.main.radarid AS
    SELECT * FROM df
""")

print(con.execute(
    "SELECT COUNT(*) FROM radarid_db.main.radarid"
).fetchone())
```

### Jagame

```python
con.execute("""
    CREATE SHARE radarid_share FROM radarid_db
    (ACCESS ORGANIZATION, VISIBILITY DISCOVERABLE, UPDATE AUTOMATIC)
""")
```

```python
print(con.execute("LIST SHARES").df())
```
